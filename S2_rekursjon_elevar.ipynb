# Del 2 — Rekursjonsprinsippet

Ikkje alle følger har ein enkel eksplisitt formel. Ofte er det lettare å beskriva **korleis du kjem frå eitt ledd til det neste**.

> ### Rekursiv (implisitt) definisjon
> Ei følge er definert **rekursivt** når beskrivinga av eit ledd inneheld eitt eller fleire av dei tidlegare ledda i følga.
>
> Ein slik definisjon treng alltid **to** delar:
> 1. ein eller fleire **startverdiar**
> 2. ein **rekursiv formel** som seier korleis du reknar deg vidare
>
> Manglar du startverdien, er følga ikkje bestemt. Manglar du formelen, kjem du ikkje vidare.

Ordet *implisitt* peikar på nettopp dette: formelen seier ikkje kva $a_n$ **er**, han seier kva $a_n$ har å gjera med naboen sin.

# Del 3 — Rekursive funksjonar

Til no har vi laga rekursive følger med **løkker**. No skal vi gjera noko som ser rart ut første gongen:

> **Vi lagar ein funksjon som kallar på seg sjølv.**

Tanken er å oversetja den rekursive formelen heilt direkte til kode. Formelen $a_n = 3a_{n-1} - 8$ blir til `return 3*a_rek(n-1) - 8`.

Men då oppstår eit problem: når stoppar det?

## Døme 3 — same følge, ny form

````python
def a_rek(n):
    if n == 1:              # rekursjonsbotn
        return 5
    else:
        return 3*a_rek(n - 1) - 8

for n in range(1, 9):
    print(f"a_{n} = {a_rek(n)}")
````

Sjekk at du får nøyaktig same tal som i døme 2. **Same følge, to ulike implementasjonar.**

> ### Rekursjonsbotn
> Testen `if n == 1` er **rekursjonsbotnen**. Det er tilfellet der funksjonen veit svaret utan å spørja seg sjølv. Utan han spør funksjonen seg sjølv i det uendelege.

## Kva skjer eigentleg?

Køyr cella under. Ho gjer det same, men skriv ut kvart einaste kall.

````python
def a_spor(n, djupn=0):
    innrykk = "    " * djupn
    print(f"{innrykk}spør etter a({n})")
    if n == 1:
        print(f"{innrykk}-> botn! a(1) = 5")
        return 5
    else:
        førre = a_spor(n - 1, djupn + 1)
        svar = 3*førre - 8
        print(f"{innrykk}-> a({n}) = 3*{førre} - 8 = {svar}")
        return svar

a_spor(5)
````

**Spørsmål.** Les utskrifta ovanfrå og ned, så nedanfrå og opp. Kva skjer i kvar retning?

<details>
<summary>🔍 Svar</summary>

**Nedover** blir heile utrekninga *utsett* — maskina kan ikkje svara på `a(5)` før ho veit `a(4)`, og så vidare heilt ned til botnen.

**Oppover** blir svara *sette saman* att.

Ei god formulering å hugsa: *rekursjon er å utsetja arbeidet til nokon endeleg kan svara.*
</details>

## Kva om botnen manglar?

Køyr cella under. Ho **skal** feila.

````python
def utan_botn(n):
    return 3*utan_botn(n - 1) - 8

try:
    utan_botn(5)
except RecursionError:
    print("RecursionError: funksjonen kalla seg sjølv for djupt.")
````

````python
def b_rek(n):
    if n == 1:
        return 1
    elif n == 2:
        return 5
    else:
        return b_rek(n - 1) + 2*b_rek(n - 2)

print([b_rek(n) for n in range(1, 11)])
````

# Del 5 — Frå rekursiv til eksplisitt

> **Dette er den viktigaste delen. Bruk tid her.**

Ei rekursiv følge og ei eksplisitt følge kan vera **den same følga**. Spørsmålet er om vi klarer å finna den eksplisitte formelen.

## Døme 5

$$a_1 = 10, \qquad a_n = 0{,}5\,a_{n-1} + 3$$

````python
d_verdiar = []
x = 10
for n in range(1, 16):
    d_verdiar.append(x)
    x = 0.5*x + 3

print([round(v, 4) for v in d_verdiar])

plt.plot(range(1, 16), d_verdiar, "o-")
plt.axhline(6, linestyle="--", color="grey")
plt.xlabel("n"); plt.ylabel("$a_n$")
plt.title("Følga nærmar seg 6")
plt.grid(True); plt.show()
````

Følga nærmar seg 6. Men kvifor akkurat 6?

Står følga heilt i ro, må $a_n = a_{n-1}$. Kall den verdien $F$:

$$F = 0{,}5F + 3 \qquad \Longrightarrow \qquad 0{,}5F = 3 \qquad \Longrightarrow \qquad F = 6$$

Dette kallar vi **fikspunktet**. Sjå no på **avstanden til fikspunktet**:

````python
avstand = [v - 6 for v in d_verdiar]
print([round(d, 5) for d in avstand])
````

**Diskuter før du går vidare.** Kva slags følge er avstandane?

# Del 6 — Kva kostar rekursjonen?

Rekursive funksjonar er elegante. Men dei kan vera **sløsande**.

Tenk på `b_rek(6)`. For å rekna han ut må maskina rekna `b_rek(5)` og `b_rek(4)`. For å rekna `b_rek(5)` må ho rekna `b_rek(4)` og `b_rek(3)` — og då reknar ho `b_rek(4)` **på nytt**. Vi teljar:

````python
tal_kall = 0

def b_teljande(n):
    global tal_kall
    tal_kall += 1
    if n == 1:
        return 1
    elif n == 2:
        return 5
    else:
        return b_teljande(n - 1) + 2*b_teljande(n - 2)

for n in [5, 10, 20, 25]:
    tal_kall = 0
    b_teljande(n)
    print(f"b({n}) krev {tal_kall} funksjonskall")
````

Ei enkel løysing: **hugs svara du alt har rekna ut.** Dette heiter *memoisering*.

````python
minne = {}

def b_hugsande(n):
    if n in minne:
        return minne[n]
    if n == 1:
        svar = 1
    elif n == 2:
        svar = 5
    else:
        svar = b_hugsande(n - 1) + 2*b_hugsande(n - 2)
    minne[n] = svar
    return svar

print(b_hugsande(60))
````

````python
t0 = time.perf_counter(); b_rek(34); t1 = time.perf_counter()
print(f"Naiv rekursjon:  {t1 - t0:.3f} sekund")

minne = {}
t0 = time.perf_counter(); b_hugsande(34); t1 = time.perf_counter()
print(f"Med memoisering: {t1 - t0:.6f} sekund")
````
